# Notebook 1 – Exploratory Data Analysis (EDA)

This notebook performs EDA on the credit card fraud dataset located in `../csv_files/creditcard.csv`.

It includes:
- Data loading and overview
- Class imbalance analysis
- Basic statistics
- Correlation analysis
- Visualisations saved to `../docs/figures/`.


```
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = os.path.join("..", "csv_files", "creditcard.csv")
FIG_DIR = os.path.join("..", "docs", "figures")
os.makedirs(FIG_DIR, exist_ok=True)

df = pd.read_csv(DATA_PATH)
df.head()
```

```
# Basic info and summary
df.info()
```

```
# Class imbalance
class_counts = df['Class'].value_counts().sort_index()
print(class_counts)
print('\nFraud ratio:', class_counts[1] / class_counts.sum())

plt.figure(figsize=(4,4))
class_counts.plot(kind='bar')
plt.title('Class Distribution (0 = Legit, 1 = Fraud)')
plt.xlabel('Class')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'class_distribution.png'), dpi=150)
plt.show()
```

```
# Amount distribution
plt.figure(figsize=(6,4))
sns.histplot(df['Amount'], bins=50, kde=False)
plt.title('Transaction Amount Distribution')
plt.xlabel('Amount')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'amount_distribution.png'), dpi=150)
plt.show()
```

```
# Correlation heatmap of PCA features and Amount
corr_cols = [c for c in df.columns if c.startswith('V')] + ['Amount', 'Class']
corr = df[corr_cols].corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'correlation_heatmap.png'), dpi=150)
plt.show()
```
